# Example 01: Chromium scRNA-seq Metadata QC（增强版）

本 notebook 加载 Chromium 单细胞元数据，统计细胞类型分布与患者分组信息。

**Phase 0 增强图型：**
- 嵌套条形图 (Level1 × Level2)
- Sunburst 环形层级图 (Level1 → Level2)
- 患者 QC 堆叠条形图

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from src.data.loaders import load_chromium_meta
from src.data.validators import validate_chromium_meta
from src.examples.config import CHROMIUM_DIR
from src.examples.metadata_qc import (
    celltype_summary,
    filter_kept_cells,
    patient_summary,
    plot_celltype_nested_bar,
    plot_celltype_sunburst,
    plot_patient_qc_bar,
)
from src.utils.plot_style import apply_style
apply_style()
```

```python
# 加载与校验
df = load_chromium_meta(CHROMIUM_DIR)
print(f"总行数: {len(df)}")
print(f"校验: {validate_chromium_meta(df) or 'OK'}")
df.head()
```

```python
# 过滤 QC 保留的细胞
kept = filter_kept_cells(df)
print(f"QC 保留: {len(kept)} / {len(df)} ({len(kept)/len(df)*100:.1f}%)")
```

```python
# 细胞类型统计
ct = celltype_summary(kept)
ct.head(15)
```

```python
# 可视化 Level1 分布
import matplotlib.pyplot as plt

level1 = ct.groupby("Level1")["count"].sum().sort_values(ascending=True)
fig, ax = plt.subplots(figsize=(8, 5))
level1.plot.barh(ax=ax)
ax.set_xlabel("Cell Count")
ax.set_title("Chromium scRNA-seq: Cell Type Distribution (Level1)")
fig.tight_layout()
plt.show()
```

```python
# 患者统计
pt = patient_summary(df)
pt
```

## Phase 0 增强可视化

```python
# 嵌套条形图 (Level1 × Level2)
fig = plot_celltype_nested_bar(ct)
plt.show()
```

```python
# Sunburst 环形层级图 (Level1 → Level2)
fig = plot_celltype_sunburst(ct)
plt.show()
```

```python
# 患者 QC 堆叠条形图
fig = plot_patient_qc_bar(pt)
plt.show()
```
