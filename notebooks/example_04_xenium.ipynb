# Example 04: Xenium Gene Panel Summary（增强版）

本 notebook 解析 Xenium 实验元信息和基因面板配置。

**Phase 0 增强图型：**
- Descriptor 构成环形图
- Source panel 条形图
- 面板组成综合 Dashboard (descriptor × source)

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from src.data.loaders import load_xenium_meta
from src.examples.config import XENIUM_DIR
from src.examples.gene_panel_summary import (
    parse_experiment_info,
    parse_targets,
    panel_stats,
    generate_report_md,
    plot_descriptor_donut,
    plot_source_bar,
    plot_panel_composition_dashboard,
)
from src.utils.plot_style import apply_style
apply_style()
```

```python
# 加载
experiment, targets = load_xenium_meta(XENIUM_DIR)

# 解析实验信息
exp_info = parse_experiment_info(experiment)
for k, v in exp_info.items():
    print(f"  {k}: {v}")
```

```python
# 解析基因面板
targets_df = parse_targets(targets)
print(f"Total targets: {len(targets_df)}")
targets_df.head(10)
```

```python
# 面板统计
stats = panel_stats(targets_df)
for k, v in stats.items():
    print(f"  {k}: {v}")
```

```python
# 生成 Markdown 报告（预览）
from IPython.display import Markdown

report = generate_report_md(exp_info, stats)
Markdown(report)
```

## Phase 0 增强可视化

```python
# Descriptor 构成环形图
import matplotlib.pyplot as plt
fig = plot_descriptor_donut(targets_df)
plt.show()
```

```python
# Source panel 条形图
fig = plot_source_bar(targets_df)
plt.show()
```

```python
# 面板组成综合 Dashboard
fig = plot_panel_composition_dashboard(targets_df, exp_info=exp_info)
plt.show()
```
